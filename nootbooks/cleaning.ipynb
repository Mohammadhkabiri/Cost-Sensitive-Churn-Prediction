```python
import pandas as pd

raw_data = pd.read_csv("../data/raw/Telco-Customer-Churn.csv")
df = raw_data.copy()
unsupervised_data = df.drop("Churn" ,axis=1)
unsupervised_data = unsupervised_data.drop("customerID", axis=1)
unsupervised_data["TotalCharges"] = pd.to_numeric(unsupervised_data["TotalCharges"], errors="coerce")
unsupervised_data["TotalCharges"] = unsupervised_data["TotalCharges"].fillna(0)
binary_cols = ["Partner", "Dependents" ,"PaperlessBilling" ,"PhoneService"]
for col in binary_cols :
    unsupervised_data[col] = unsupervised_data[col].map({"Yes" : 1 , "No" : 0})
categorical_cols = [
    'MultipleLines', 'InternetService', 'OnlineSecurity', 
    'OnlineBackup', 'DeviceProtection', 'TechSupport', 
    'StreamingTV', 'StreamingMovies', 'Contract', 'PaymentMethod'
]
encoded_data = pd.get_dummies(unsupervised_data, columns=categorical_cols, drop_first=True, dtype=int)
encoded_data['gender'] = encoded_data['gender'].map({'Male': 1, 'Female': 0})
encoded_data["Total_Services"] = (
    encoded_data["OnlineSecurity_Yes"] + 
    encoded_data["OnlineBackup_Yes"] + 
    encoded_data["DeviceProtection_Yes"] + 
    encoded_data["TechSupport_Yes"] + 
    encoded_data["StreamingTV_Yes"] + 
    encoded_data["StreamingMovies_Yes"]
)
encoded_data.to_csv("../data/proccessed/encoded-data.csv", index=False)
# encoded_data.info()
encoded_data.describe()

```
